Central Tendencies:
1. Mean
2. Median
3. Mode

```python
# analyse the distribution of data in the salary
fig, ax = plt.subplots(figsize=(8,8))
sns.histplot(dataset.salary, kde=True) # Histogram Plot
```

Replace the missing values with Median value

```python
dataset['salary'] = dataset['salary'].fillna(dataset['salary'].median())
# fillna = fill NaN (missing) values / fill not available
```
