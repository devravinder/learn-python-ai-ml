```python
import numpy as np
import pandas as pd
import sklearn.datasets
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
# loading the dataset
dataset = sklearn.datasets.load_breast_cancer()
```

```python
dataset
```

```python
# loading the data to a pandas dataframe
df = pd.DataFrame(dataset.data, columns=dataset.feature_names)
```

```python
df.head()
```

```python
df.shape
```

```python
X = df # for input
Y = dataset.target # for output / result 
```

```python
print(X)
```

```python
print(X.min())
```

```python
print(X.max())
```

```python
print(Y)
```

Splitting the data into training data and test data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=3)
```

`random_state` controls how the data is randomly shuffled and split.


- Without random_state ( on every run ) → different train/test data
- with random_state ( on every run ) → same random train/test split ( A fixed number used to get ).



`random_state=3 means:`

- Shuffle the data randomly
- You will get the same train/test split every time you run the code
- Use 3 as the starting point for that randomness
- 3 has no special meaning. we can use random_state=1, random_state=10 etc

## Standardize the data

```python
print(dataset.data.std())
```

```python
scaler = StandardScaler()
```

```python
scaler.fit(X_train)
```

```python
X_train_standardized = scaler.transform(X_train)
```

### After standardization

```python
print(X_train_standardized.min()) # overall
```

```python
print(X_train_standardized.max())
```

```python
print(X_train_standardized.min(axis=0)) # `axis=0` means: perform the operation `column-wise`.
print(X_train_standardized.max(axis=0))
```
